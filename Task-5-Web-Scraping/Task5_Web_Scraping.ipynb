# Task 5: Web Data Extraction & Analysis

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://books.toscrape.com/"
response = requests.get(url)
```

```python
response.status_code
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
books = soup.find_all('article', class_='product_pod')
len(books)
```

```python
books[0].h3.a['title']
```

```python
books[0].find('p', class_='price_color').text
```

```python
books[0].find('p', class_='star-rating')['class'][1]
```

```python
titles = [book.h3.a['title'] for book in books]
```

```python
prices = [book.find('p', class_='price_color').text for book in books]
```

```python
ratings = [book.find('p', class_='star-rating')['class'][1] for book in books]
```

```python
scraped_data = pd.DataFrame({
    'Book_Title': titles,
    'Price': prices,
    'Rating': ratings
})
```

```python
scraped_data['Price'] = (
    scraped_data['Price']
    .str.replace('Â£', '', regex=False)
    .str.replace('£', '', regex=False)
    .astype(float)
)
```

```python
rating_map = {
    'One': 1,
    'Two': 2,
    'Three': 3,
    'Four': 4,
    'Five': 5
}

scraped_data['Rating'] = scraped_data['Rating'].map(rating_map)
```
